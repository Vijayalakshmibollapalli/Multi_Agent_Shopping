## AI Shopping Intelligence

#### Multi-Agent Product Research and Purchase Decision Assistant using LangGraph and MCP

#### 1. Project Overview

AI Shopping Intelligence is an AI-powered shopping assistant that researches products based on a user's natural-language request and provides an evidence-based purchase recommendation.

The system is not limited to specific product categories. Users can search for laptops, phones, headphones, cameras, monitors, furniture, appliances, accessories, or any other products.

#### 2. Main Technologies
* Python
* LangGraph
* LangChain
* Groq LLM
* MCP (Model Context Protocol)
* FastMCP
* Tavily Search
* FastAPI
* Frontend: HTML, CSS, JavaScript

### 3. How It Works

````python
User Shopping Query
        ↓
Requirements Analysis
        ↓
LangGraph Workflow
        ↓
MCP Server
   ┌────┼─────┬──────────┐
   ↓    ↓     ↓          ↓
Products Prices Reviews Comparisons
   └────┼─────┴──────────┘
        ↓
Evidence-Based Recommendation
        ↓
FastAPI Web Interface
````

### 4. MCP Server — mcp_server.py

The MCP server connects the application with Tavily web search.

It provides four tools:

* search_products → Finds real products and specifications.
* search_prices → Finds current prices and availability.
* search_reviews → Finds reviews, pros, cons, and feedback.
* search_comparison → Finds alternatives and product comparisons.

The server is deployed remotely using FastMCP.

#### 5. Shopping Agent — shopping_agent.py

The shopping agent is the main intelligence layer.

It uses LangGraph to create the workflow:

Requirements
     ↓
Research
     ↓
Recommendation

The agent:

1) Understands the user's shopping requirements.
2) Extracts budget, purpose, preferences, and important features.
3) Calls MCP research tools.
4) Collects product, price, review, and comparison evidence.
5) Sends the relevant evidence to the Groq LLM.
6) Generates a concise recommendation.
7) Verifies price, specifications, and reviews.

#### 7. Web Interface — app.py

app.py uses FastAPI to provide a simple interactive interface.

The user enters a natural-language query such as:

`Best laptop under ₹70000 for programming`

or

`Best office chair under ₹20000 for long working hours`

The application sends the query to the LangGraph shopping agent and displays the recommendation.

### How the Project Works

#### Step 1: User enters a query

The user enters a natural-language shopping request.

Example:

```text
Best headphones under ₹5,000
```

#### Step 2: Requirement analysis

The LangGraph agent analyzes the query and identifies important requirements such as:

* Product type → Headphones
* Budget → ₹5,000
* Purpose → Based on the user's request
* Important features → Based on the user's request

#### Step 3: Product research

The MCP server uses Tavily to search the web for real products and their specifications.

#### Step 4: Price research

The MCP server searches for current prices and availability of the products.

#### Step 5: Review research

The system searches for expert reviews, user feedback, pros, cons, and product problems.

#### Step 6: Product comparison

The system searches for similar products and alternatives that may match the user's requirements.

#### Step 7: AI analysis

LangGraph collects the research results and sends the relevant evidence to the Groq LLM.

The AI analyzes the evidence, checks the user's requirements and budget, and selects a suitable product.

#### Step 8: Final result

The user receives a concise recommendation containing:

* Recommended choice
* Current price
* Why it matches
* Key details
* Pros
* Cons
* Verification status
* Alternatives
* Final recommendation
